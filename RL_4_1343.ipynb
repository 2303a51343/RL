Upper Confidence Bound

```python
import math

machines = ['A', 'B', 'C']
N_play = 10

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': [], 'B': [], 'C': []}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}
UCB = {'A': 0.0, 'B': 0.0, 'C': 0.0}

total_reward = 0

print("UCB Multi-Armed Bandit")

for machine in machines:
    print("\nPlay with machine", machine)
    reward = float(input("Enter reward: "))

    R[machine].append(reward)
    total_reward += reward

    N[machine] += 1

    Q[machine] = Q[machine] + (reward - Q[machine]) / N[machine]

for t in range(3, N_play):

    print("\nPlay number:", t + 1)

    for machine in machines:
        UCB[machine] = Q[machine] + math.sqrt(
            (2 * math.log(t + 1)) / N[machine]
        )

    print("UCB values:")
    for machine in machines:
        print(machine, "=", round(UCB[machine], 3))

    selected = max(UCB, key=UCB.get)

    print("Play with machine", selected)
    reward = float(input("Enter reward: "))

    R[selected].append(reward)
    total_reward += reward

    N[selected] += 1

    Q[selected] = Q[selected] + (
        reward - Q[selected]
    ) / N[selected]


for machine in machines:
    UCB[machine] = Q[machine] + math.sqrt(
        (2 * math.log(N_play)) / N[machine]
    )

print("\n--- FINAL RESULTS ---")
print("Machine\tN\tRewards\t\tUCB\tQ-value")

for machine in machines:
    print(
        machine,
        "\t",
        N[machine],
        "\t",
        R[machine],
        "\t",
        round(UCB[machine], 3),
        "\t",
        round(Q[machine], 3)
    )

print("\nTotal Reward Received =", total_reward)
```
