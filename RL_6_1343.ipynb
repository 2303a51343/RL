```python
import gymnasium as gym
def random_agent(environment_name, episodes=5):

    env = gym.make(environment_name)
    print("\nEnvironment:", environment_name)
    print("--------------------------------")

    episode_rewards = []
    for episode in range(1, episodes + 1):

        state, info = env.reset()
        total_reward = 0
        steps = 0

        terminated = False
        truncated = False

        while not (terminated or truncated):
            action = env.action_space.sample()
            next_state, reward, terminated, truncated, info = env.step(action)

            total_reward += reward
            steps += 1

            state = next_state

        average_reward = total_reward / steps

        episode_rewards.append(total_reward)

        print("Episode", episode)
        print("Total Reward:", total_reward)
        print("Steps:", steps)
        print("Average Reward:", round(average_reward, 2))
        print()
    overall_average = sum(episode_rewards) / len(episode_rewards)

    print("Total Rewards:", episode_rewards)
    print("Overall Average Reward:", round(overall_average, 2))

    env.close()

random_agent("CliffWalking-v1", episodes=5)
random_agent("Taxi-v4", episodes=5)
```
